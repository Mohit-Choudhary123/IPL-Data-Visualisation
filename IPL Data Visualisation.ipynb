Which team won the most matches


```python
match_winner=df['winner'].value_counts()
sns.barplot(y=match_winner.index,x=match_winner.values,palette='viridis')
plt.title('Most match win by a team')
plt.xlabel("no. of matches")
plt.ylabel('teams')
```
